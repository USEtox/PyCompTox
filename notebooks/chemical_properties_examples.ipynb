# Chemical Properties Examples

This notebook demonstrates how to use the `ChemicalProperties` class to retrieve:
- Property summaries (physchem and environmental fate)
- Predicted properties from QSAR models
- Experimental (measured) properties
- Environmental fate and transport data

## Setup

First, make sure you have your API key configured.

```python
from pycomptox.chemical import Chemical, ChemicalProperties, ChemicalDetails
import pandas as pd

# Initialize clients
searcher = Chemical()
props = ChemicalProperties()
details = ChemicalDetails()

print("✓ Clients initialized successfully")
```

## Example 8: Batch Operations

Get properties for multiple chemicals efficiently.

```python
# Search for multiple chemicals
chemical_names = ["Caffeine", "Aspirin", "Ibuprofen"]
dtxsids = []

for name in chemical_names:
    results = searcher.search_by_exact_value(name)
    if results:
        dtxsid = results[0]['dtxsid']
        dtxsids.append(dtxsid)
        print(f"✓ {name}: {dtxsid}")

print(f"\nRetrieving predicted properties for {len(dtxsids)} chemicals...")

# Get batch predicted properties
batch_predicted = props.get_predicted_properties_by_dtxsid_batch(dtxsids)

print(f"Retrieved {len(batch_predicted)} total property predictions\n")

# Group by chemical
df_batch = pd.DataFrame(batch_predicted)
print("Properties per chemical:")
print(df_batch.groupby('dtxsid').size())

# Show sample
print("\nSample data:")
columns = ['dtxsid', 'propName', 'propValue', 'propUnit']
display(df_batch[columns].head(15))
```

```python
# The batch method now works correctly!
# You can retrieve properties for multiple chemicals efficiently

print(f"DTXSIDs being queried: {dtxsids}")
print(f"Number of chemicals: {len(dtxsids)}")
print()

# Test the batch method
batch_result = props.get_predicted_properties_by_dtxsid_batch(dtxsids)
print(f"Retrieved {len(batch_result)} total predicted properties")
print()

# Count properties per chemical
from collections import Counter
per_chemical = Counter(r['dtxsid'] for r in batch_result)
print("Properties per chemical:")
for dtxsid, count in per_chemical.items():
    print(f"  {dtxsid}: {count} properties")
```

### Batch Method Fix

The batch methods (`get_predicted_properties_by_dtxsid_batch`, `get_experimental_properties_by_dtxsid_batch`, and `get_fate_by_dtxsid_batch`) have been fixed to work correctly with the API. They now:
- Use the proper `_make_cached_request` method
- Support caching with the `use_cache` parameter
- Follow the same patterns as other API methods in the package

## Example 9: Compare Predicted vs Experimental

Compare predicted QSAR values with experimental measurements.

```python
# Use Bisphenol A
dtxsid = "DTXSID7020182"

# Get both predicted and experimental
predicted = props.get_predicted_properties_by_dtxsid(dtxsid)
experimental = props.get_experimental_properties_by_dtxsid(dtxsid)

# Create DataFrames
df_pred = pd.DataFrame(predicted)
df_exp = pd.DataFrame(experimental)

print(f"Predicted properties: {len(df_pred)}")
print(f"Experimental measurements: {len(df_exp)}")
print()

# Find common properties
common_props = set(df_pred['propName'].unique()) & set(df_exp['propName'].unique())
print(f"Properties with both predicted and experimental data: {len(common_props)}")
print("\nCommon properties:")
for prop in sorted(common_props)[:10]:
    print(f"  - {prop}")
```

## Example 10: Complete Workflow

From search to comprehensive property analysis.

```python
def analyze_chemical(chemical_name):
    """Complete property analysis for a chemical."""
    print(f"Analyzing: {chemical_name}")
    print("="*70)
    
    # 1. Search
    results = searcher.search_by_exact_value(chemical_name)
    if not results:
        print(f"Chemical not found: {chemical_name}")
        return
    
    dtxsid = results[0]['dtxsid']
    print(f"✓ Found: {results[0]['preferredName']} ({dtxsid})")
    print()
    
    # 2. Property Summary
    summary = props.get_property_summary_by_dtxsid(dtxsid)
    print(f"Property Summaries: {len(summary)}")
    
    # 3. Predicted Properties
    predicted = props.get_predicted_properties_by_dtxsid(dtxsid)
    print(f"Predicted Properties: {len(predicted)}")
    
    # 4. Experimental Properties
    experimental = props.get_experimental_properties_by_dtxsid(dtxsid)
    print(f"Experimental Measurements: {len(experimental)}")
    
    # 5. Fate Properties
    fate = props.get_fate_summary_by_dtxsid(dtxsid)
    print(f"Fate Properties: {len(fate)}")
    print()
    
    # Show some key properties
    if summary:
        print("Key Property Summaries:")
        df = pd.DataFrame(summary)
        for _, row in df.head(5).iterrows():
            print(f"  {row['propName']}: ", end="")
            if pd.notna(row.get('experimentalMedian')):
                print(f"{row['experimentalMedian']} {row.get('unit', '')} (exp)")
            elif pd.notna(row.get('predictedMedian')):
                print(f"{row['predictedMedian']} {row.get('unit', '')} (pred)")
            else:
                print("N/A")
    
    print("\n" + "="*70 + "\n")

# Analyze several chemicals
chemicals = ["Bisphenol A", "Caffeine", "Aspirin"]
for chem in chemicals:
    analyze_chemical(chem)
```
